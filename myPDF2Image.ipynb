```python
# AI from PDF to image

#from pdf2image import convert_from_path
#import os

def convert_pdf_to_images(pdf_path, output_folder, dpi=200, output_format='png'):
    """
    Converts each page of a PDF to an image.

    Args:
        pdf_path (str): Path to the PDF file.
        output_folder (str): Path to the output folder for images.
        dpi (int): Resolution of the output images in DPI.
        output_format (str): Format of the output images (e.g., 'png', 'jpeg').
    """
    
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    images = convert_from_path(pdf_path, dpi=dpi)

    for i, image in enumerate(images):
        image_path = os.path.join(output_folder, f'page_{i+1}.{output_format}')
        image.save(image_path)
    print("Conversion completed")

# Example usage:
pdf_file = "samples/DTW_2190599060_ANE_diGLmNgD3SlTBEw8wLOR8O0w2aCO53Spjdh.pdf"
output_dir = 'samples'

convert_pdf_to_images(pdf_file, output_dir)
```

### Extract images from .pdfs

```python
import fitz
import io
from PIL import Image
```

```python
# open the file
FILE_NAME = "PDFs/DTW_2190599060_ANE_diGLmNgD3SlTBEw8wLOR8O0w2aCO53Spjdh.pdf"
pdf_file = fitz.open(FILE_NAME)
```

```python
# iterate over PDF pages
for page_index in range(len(pdf_file)):
    # get the page itself
    page = pdf_file.load_page(page_index)  # load the page
    image_list = page.get_images(full=True)  # get images on the page

    # printing number of images found in this page
    if image_list:
        print(f"[+] Found a total of {len(image_list)} images on page {page_index}")
    else:
        print("[!] No images found on page", page_index)
    
    for image_index, img in enumerate(image_list, start=1):
        # get the XREF of the image
        xref = img[0]

        # extract the image bytes
        base_image = pdf_file.extract_image(xref)
        image_bytes = base_image["image"]

        # get the image extension
        image_ext = base_image["ext"]

        # save the image
        image_name = f"image{page_index+1}_{image_index}.{image_ext}"
        with open(image_name, "wb") as image_file:
            image_file.write(image_bytes)
            print(f"[+] Image saved as {image_name}")
```
